## Step 2: Import Libraries

Here we load all the tools we need:
- `os` — helps manage system settings
- `userdata` — lets us safely use our secret API key stored in Colab
- `Groq` — the main library to connect with the AI

```python
import os
from google.colab import userdata
from groq import Groq
```

## Step 3: Setup Groq Client

We need an **API Key** to connect with Groq. Think of it as a password that gives us access to the AI.

### 🛠️ How to get your FREE Groq API Key:
1. Go to 👉 [https://console.groq.com](https://console.groq.com)
2. Sign up for a free account
3. Click on **"API Keys"** from the left sidebar
4. Click **"Create API Key"** and copy it

### 🔐 How to add it in Colab:
1. Click the **🔑 Secrets icon** on the left sidebar in Colab
2. Click **"Add new secret"**
3. Name it exactly: `Groq_Api_Key`
4. Paste your API key in the value field
5. Toggle **"Notebook access"** to ON

```python
client = Groq(
    api_key=userdata.get("Groq_Api_Key"),
)
```

## Step 4: Ask Your Question

This cell will ask you to type a question. Whatever you type will be sent to the AI model.

```python
user_query = input("What do you want to search? 👉 ")
```

## Step 5: Get AI Response

We send your question to the **LLaMA 3.3 70B** model (a powerful AI made by Meta) and print the answer below.

```python
chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "user",
            "content": user_query,
        }
    ],
    model="llama-3.3-70b-versatile",
)

print(chat_completion.choices[0].message.content)
```
